## Q2

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print(arr[::-1])
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
val_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = val_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]
print(most_freq_x)
print(indices_x)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
val_y, counts_y = np.unique(y, return_counts=True)
most_freq_y_vals = val_y[counts_y == counts_y.max()]
for val in most_freq_y_vals:
    indices_y = np.where(y == val)[0]
    print(val)
    print(indices_y)
```

## Q4

```python
deepak = np.linspace(10, 100, 25)
print(deepak.ndim)
print(deepak.shape)
print(deepak.size)
print(deepak.dtype)
print(deepak.nbytes)

transpose_reshape = deepak.reshape(25, 1)
print(transpose_reshape)

print(deepak.T)
```

## Q5

```python
ucs420_deepak = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print(np.mean(ucs420_deepak))
print(np.median(ucs420_deepak))
print(np.max(ucs420_deepak))
print(np.min(ucs420_deepak))
print(np.unique(ucs420_deepak))

reshaped_ucs420_deepak = ucs420_deepak.reshape(4, 3)
print(reshaped_ucs420_deepak)

reshaped_ucs420_deepak = np.resize(ucs420_deepak,(2, 3))
print(reshaped_ucs420_deepak)
```
