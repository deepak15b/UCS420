### 1. Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
import random
random.seed(1024170432)
numbers = [random.randint(100,900) for _ in range(100)]
print(numbers)
```

### 2. Count and print all odd numbers in the list.

```python
odd_numbers = [i for i in numbers if i % 2 != 0]
print("Total odd numbers: ",len(odd_numbers))
print(odd_numbers)
```

### 3. count and print all the even numbers


```python
even_numbers = [x for x in numbers if x % 2 == 0]
print("Total even numbers:",len(even_numbers))
print(even_numbers)
```

### 4.Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = [x for x in numbers if is_prime(x)]
print("Prime numbers:", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))
```

### 5. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
frequency = {x: numbers.count(x) for x in set(numbers)}

most_frequent = max(frequency, key=frequency.get)
count = frequency[most_frequent]

print("\nMost frequent number:", most_frequent)
print("It occurs:", count, "time(s)")
```

## 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 foreach digit} and set B = {digit × 9 for each digit}. Print both sets.

```python
digits = [1,0,2,4,1,7,0,4,3,2]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)
```

### vi.  Find the union of A and B — all unique values across both sets.

```python
union_set = A.union(B)
print("A union B:", union_set)
```

### Vii. Find the intersection of A and B — values common to both.

```python
intersection_set = A.intersection(B)

print("Intersection of A and B:", intersection_set)
```

### Viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how thisdiffers from symmetric_difference().

```python
## A- B means values present in A but not in B 
# B - A means values in B but not in B

A_minus_B = A.difference(B)

print("A - B:", A_minus_B)

B_minus_A = B.difference(A)

print("B - A:", B_minus_A)
```

### ix. Find the symmetric difference of A and B.

```python
symmetric = A.symmetric_difference(B)

print("Symmetric difference:", symmetric)
```

### x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

### xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists.

```python
x = int(input("Enter a value to remove from A: "))

A.discard(x)

print("Set A after discard:", A)

# Why discard() instead of remove() 
# beacuse if we use remove() and any key is not present it will give key error 
# if we use discard() even if the value is not present it will not give any error
```
